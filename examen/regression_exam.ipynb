```python
import numpy as np
import pandas as pd
```

```python
# 1. CARGA DE DATOS
data = pd.read_csv('../data/sales_data.csv')

x = data['advertising'].to_numpy()
y = data['sales'].to_numpy()

print(data)
print(x)
print(y)
```

```python
# 2. MATRIZ DE DISEÑO
X = np.column_stack((np.ones(len(x)), x))

print('X:')
print(X)
print('Shape X:', X.shape)
print('Shape y:', y.shape)
```

```python
# 3. OPERACIONES MATRICIALES
XtX = X.T @ X
Xty = X.T @ y

print('X.T @ X:')
print(XtX)
print('X.T @ y:')
print(Xty)
```

```python
# 4. ESTIMACION DE PARAMETROS
B, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)
beta = B

beta_0 = beta[0]
beta_1 = beta[1]

print('Beta 0:', beta_0)
print('Beta 1:', beta_1)
```

```python
# 5. PREDICCION
x_new = np.array([1, 9])
prediction = x_new @ beta

print('Prediction:', prediction)
```

```python
# 6. PREDICCIONES DEL DATASET
y_pred = X @ beta

print(y_pred)
```

```python
# 7. ERROR
errors = y - y_pred
error_norm = np.linalg.norm(errors)

print('Error vector:')
print(errors)
print('Error norm:')
print(error_norm)
```
