```python
import numpy as np

# Datos
x = np.array([2, 4, 6, 8, 10], dtype=float)
y = np.array([5.0, 7.1, 9.2, 10.8, 13.1], dtype=float)

# 1. Vector x
print("x =", x)

# 2. Vector y
print("y =", y)

# 3. Matriz X con columna de unos
X = np.column_stack((np.ones(len(x)), x))
print("X =\n", X)

# 4. X^T X
XTX = X.T @ X
print("X^T X =\n", XTX)

# 5. X^T y
XTy = X.T @ y
print("X^T y =", XTy)

# 6. Vector beta_hat
beta_hat = np.linalg.inv(XTX) @ XTy
print("beta_hat =", beta_hat)

# 7. Predicciones y_hat
Y_hat = X @ beta_hat
print("y_hat =", Y_hat)

# 8. Vector de errores
errors = y - Y_hat
print("errores =", errors)

# 9. Norma del error
error_norm = np.linalg.norm(errors)
print("norma del error =", error_norm)

# 10. Prediccion para x = 12
x_new = 12
prediction = beta_hat[0] + beta_hat[1] * x_new
print("Prediccion para 12 productos =", prediction)
```
